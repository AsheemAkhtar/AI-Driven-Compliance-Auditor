# IMPORTING LIBRARIES

```python
import fitz
import numpy as np
import pickle
import ollama
from pathlib import Path
from langchain_text_splitters import RecursiveCharacterTextSplitter
from sentence_transformers import SentenceTransformer
from sentence_transformers.util import cos_sim
```

```python
RAW_FOLDER = Path("DATA/RAW")
```

# FETCHING THE DATA

```python
pdf_files = list(RAW_FOLDER.glob("*.pdf"))

print(f"Found {len(pdf_files)} PDF files\n")

for pdf in pdf_files:
    print(pdf.name)
```

```python
sample_pdf = pdf_files[0]

doc = fitz.open(sample_pdf)

text = ""

for page in doc:
    text += page.get_text()

print(text[:250])
```

```python
documents = []

for pdf_path in pdf_files:
    
    doc = fitz.open(pdf_path)
    
    full_text = ""
    
    for page in doc:
        full_text += page.get_text()
    
    documents.append({
        "filename": pdf_path.name,
        "text": full_text
    })

print(f"Processed {len(documents)} documents")
```

# CHECKING DATA OF EACH DOCUMENT

```python
for doc in documents:
    print("=" * 60)
    print(doc["filename"])
    print("-" * 60)
    print(doc["text"][:500])
    print("\n")
```

```python
for doc in documents:
    print(
        f"{doc['filename']} --> "
        f"{len(doc['text'])} characters"
    )
```

# CREATING CHUNKS

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=150
)
```

```python
all_chunks = []

for doc in documents:
    
    chunks = text_splitter.split_text(doc["text"])
    
    for i, chunk in enumerate(chunks):
        
        all_chunks.append({
            "filename": doc["filename"],
            "chunk_id": i,
            "text": chunk
        })

print(f"Created {len(all_chunks)} chunks")
```

```python
sample_chunk = all_chunks[0]

print("FILENAME:\n")
print(sample_chunk["filename"])

print("\nCHUNK ID:\n")
print(sample_chunk["chunk_id"])

print("\nTEXT:\n")
print(sample_chunk["text"])
```

# EMBEDDING THE TEXT

```python
embedding_model = SentenceTransformer("BAAI/bge-m3")
```

```python
sample_text = all_chunks[50]["text"]

embedding = embedding_model.encode(sample_text)

print(type(embedding))
print(embedding.shape)
```

```python
chunk_texts = [chunk["text"] for chunk in all_chunks]

chunk_embeddings = embedding_model.encode(
    chunk_texts,
    show_progress_bar=True
)
```

# QUERING THE KNOWLEDGE BASE

```python
def semantic_search(query, top_k=3):
    
    query_embedding = embedding_model.encode(query)
    
    similarities = cos_sim(
        query_embedding,
        chunk_embeddings
    )[0]
    
    top_results = similarities.argsort(descending=True)[:top_k]
    
    for rank, idx in enumerate(top_results):
        
        score = similarities[idx].item()
        
        print("=" * 80)
        print(f"RESULT #{rank+1}")
        print(f"SCORE: {score:.4f}")
        print("=" * 80)
        
        print(f"\nFILENAME: {all_chunks[idx]['filename']}\n")
        
        print(all_chunks[idx]["text"][:1200])
        print("\n")
```

```python
semantic_search("What are customer due diligence requirements?")
```

# UPLOADING THE DOCUMENT WHICH NEEDED TO BE COMPARED WITH THE KNOWLEDGE BASE

```python
UPLOAD_FOLDER = Path("DATA/UPLOADS")

uploaded_files = list(UPLOAD_FOLDER.glob("*.pdf"))

print(uploaded_files)
```

```python
uploaded_docs = []

for pdf_path in uploaded_files:
    
    doc = fitz.open(pdf_path)
    
    full_text = ""
    
    for page in doc:
        full_text += page.get_text()
    
    uploaded_docs.append({
        "filename": pdf_path.name,
        "text": full_text
    })

print(f"Processed {len(uploaded_docs)} uploaded documents")
```

```python
print(uploaded_docs[0]["text"][:3000])
```

# CREATING CHUNKS

```python
uploaded_chunks = []

for doc in uploaded_docs:
    
    chunks = text_splitter.split_text(doc["text"])
    
    for i, chunk in enumerate(chunks):
        
        uploaded_chunks.append({
            "filename": doc["filename"],
            "chunk_id": i,
            "text": chunk
        })

print(f"Created {len(uploaded_chunks)} uploaded chunks")
```

```python
uploaded_texts = [
    chunk["text"]
    for chunk in uploaded_chunks
]

uploaded_embeddings = embedding_model.encode(
    uploaded_texts,
    batch_size=16,
    show_progress_bar=True
)
```

# ASKING QUESTIONS ABOUT THE UPLOADED DOCUMENT

```python
def search_uploaded_docs(query, top_k=3):
    
    query_embedding = embedding_model.encode(query)
    
    similarities = cos_sim(
        query_embedding,
        uploaded_embeddings
    )[0]
    
    top_results = similarities.argsort(descending=True)[:top_k]
    
    results = []
    
    for idx in top_results:
        
        results.append({
            "score": similarities[idx].item(),
            "filename": uploaded_chunks[idx]["filename"],
            "text": uploaded_chunks[idx]["text"]
        })
    
    return results
```

```python
search_uploaded_docs(
    "أWhich section discusses Human rights?"
)
```

```python
response = ollama.chat(
    model="qwen2.5:7b",
    messages=[
        {
            "role": "user",
            "content": "What is AML compliance"
        }
    ]
)

print(response["message"]["content"])
```

```python
def build_context(results):
    
    context = ""
    
    for i, result in enumerate(results):
        
        context += f"""
DOCUMENT CHUNK {i+1}
SOURCE: {result['filename']}

{result['text']}

==================================================
"""
    
    return context
```

```python
def ask_uploaded_document(query, top_k=3):
    
    results = search_uploaded_docs(
        query,
        top_k=top_k
    )
    
    context = build_context(results)
    
    prompt = f"""
You are a compliance AI assistant.

Answer the user's question ONLY using the provided context.

If the answer is not found in the context, say:
"I could not find the answer in the uploaded document."

USER QUESTION:
{query}

CONTEXT:
{context}

ANSWER:
"""
    
    response = ollama.chat(
        model="qwen2.5:7b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )
    
    return response["message"]["content"]
```

```python
answer = ask_uploaded_document(
    "Which section discusses human rights?"
)

print(answer)
```

```python
def search_regulations(query, top_k=3):
    
    query_embedding = embedding_model.encode(query)
    
    similarities = cos_sim(
        query_embedding,
        chunk_embeddings
    )[0]
    
    top_results = similarities.argsort(descending=True)[:top_k]
    
    results = []
    
    for idx in top_results:
        
        results.append({
            "score": similarities[idx].item(),
            "filename": all_chunks[idx]["filename"],
            "text": all_chunks[idx]["text"]
        })
    
    return results
```

```python
reg_results = search_regulations(
    "customer due diligence requirements"
)

reg_results[0]
```

# CHECKING IF THE UPLOADED DOCUMENT IS COMPLIANT

```python
def compliance_check(query, top_k=3):
    
    regulation_results = search_regulations(
        query,
        top_k=top_k
    )
    
    policy_results = search_uploaded_docs(
        query,
        top_k=top_k
    )
    
    regulation_context = build_context(
        regulation_results
    )
    
    policy_context = build_context(
        policy_results
    )
    
    prompt = f"""
You are an expert AML and compliance auditor.

Your task is to compare the uploaded enterprise policy
against the regulatory requirements.

Identify:
1. Requirements mentioned in regulations
2. Whether the uploaded policy addresses them
3. Potential gaps or missing controls
4. Compliance observations

========================
REGULATORY REQUIREMENTS
========================

{regulation_context}

========================
UPLOADED POLICY
========================

{policy_context}

========================
USER REQUEST
========================

{query}

========================
ANSWER
========================
"""
    
    response = ollama.chat(
        model="qwen2.5:7b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )
    
    return response["message"]["content"]
```

# AUDIT RESULTS

```python
answer = compliance_check(
    "Does the uploaded policy address customer due diligence requirements?"
)

print(answer)
```
